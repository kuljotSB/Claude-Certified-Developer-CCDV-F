### Setting up the Environment

```python
import os
from dotenv import load_dotenv

load_dotenv()

claude_model_name = os.getenv("CLAUDE_MODEL_NAME")
claude_api_key = os.getenv("CLAUDE_API_KEY")
```

### Creating the Anthropic Client

```python
import anthropic

client = anthropic.Anthropic(api_key=claude_api_key)
```

### Configure the Memory Tool

```python
import anthropic
from anthropic.tools import BetaLocalFilesystemMemoryTool

client = anthropic.Anthropic(api_key=claude_api_key)

memory = BetaLocalFilesystemMemoryTool(
    base_path="./memory"
)
```

### Load the First Conversation

```python
runner = client.beta.messages.tool_runner(
    model=claude_model_name,
    max_tokens=1024,
    tools=[memory],
    messages=[
        {
            "role": "user",
            "content": """
                        Remember the following information.
                        My name is Kuljot Singh Bakshi. I am 21 years old and a Udemy Instructor Partner.
                        
                        My company is CarbonOps.
                        Our new AI Product is ESGOps - an AI Powered ESG Audit and Reporting Tool for BRSR, GRI, ESRS, SDGs etc.

                        Whenever you write LinkedIn posts,
                        use a professional and educational tone.

                        My preferred audience is sustainability consultants.
                        """
        }
    ]
)

runner.until_done()
```

### Start a new Conversation

```python
runner = client.beta.messages.tool_runner(
    model=claude_model_name,
    max_tokens=1024,
    tools=[memory],
    messages=[
        {
            "role": "user",
            "content": "Write a LinkedIn post announcing our new AI product."
        }
    ]
)

response = runner.until_done()

for block in response.content:
    if block.type == "text":
        print(block.text)
```
