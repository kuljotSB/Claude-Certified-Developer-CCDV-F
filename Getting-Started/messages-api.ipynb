### Setting up the Environment

```python
import os
from dotenv import load_dotenv

load_dotenv()

claude_model_name = os.getenv("CLAUDE_MODEL_NAME")
claude_api_key = os.getenv("CLAUDE_API_KEY")
```

### Creating the Anthropic Client

```python
import anthropic

client = anthropic.Anthropic(api_key=claude_api_key)

message = client.messages.create(
    model=claude_model_name,
    system = "You are Batman, the Dark Knight and Protector of Gotham City.",
    messages=[
        {
            "role": "user",
            "content": "How is Gotham City doing today?",
        }
    ],
    max_tokens = 20000
)

for block in message.content:
    if block.type == "text":
        print(block.text)
```

### Stream Messages for Better User Experience

```python
with client.messages.stream(
    model=claude_model_name,
    system = "You are Batman, the Dark Knight and Protector of Gotham City.",
    messages=[
        {
            "role": "user",
            "content": "How is Gotham City doing today?",
        }
    ],
    max_tokens = 20000
) as stream:
    for text in stream.text_stream:
        print(text, end="", flush=True)
```

### Pass Assistant Response to the Next Prompt

```python
with client.messages.stream(
    model=claude_model_name,
    system = "You are Batman, the Dark Knight and Protector of Gotham City.",
    messages=[
        {
            "role": "user",
            "content": "Hi I am Kuljot. How is Gotham City doing today?",
        },
        {
            "role": "assistant",
            "content": "Hello Kuljot! Nice to meet you. Gotham City is facing a surge in criminal activity, with the Joker and his gang causing chaos in the streets. The citizens are on edge, and the police force is stretched thin. Batman is working tirelessly to restore order and protect the innocent."
        },
        {
            "role": "user",
            "content": "Really? What is the Joker up to now? Also do you remember my name?",
        }
    ],
    max_tokens = 20000
) as stream:
    for text in stream.text_stream:
        print(text, end="", flush=True)
```
